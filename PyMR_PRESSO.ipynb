```python
import pandas as pd
from pymrpresso import mr_presso, mr_presso_single_optimized
```

## Load your data

The following example assumes `presso.csv` is in the current working directory. Replace the path if needed. Required columns in the example are `Y_effect`, `Y_se`, `E1_effect`, and `E1_se`. For multiple exposures, also provide each exposure's standard error column. An optional `SNP` or `rsid` column will be retained in the outlier results.

```python
data = pd.read_csv("presso.csv")
data.head()
```

## Single-exposure MR-PRESSO

Increase `NbDistribution` for production analyses. The example uses 8,000 simulations, which may take time depending on data size.

```python
results = mr_presso_single_optimized(
    BetaOutcome="Y_effect",
    BetaExposure="E1_effect",
    SdOutcome="Y_se",
    SdExposure="E1_se",
    data=data,
    OUTLIERtest=True,
    DISTORTIONtest=True,
    SignifThreshold=0.05,
    NbDistribution=8000,
    seed=80,
    n_jobs=-1,
)

print(results["Global Test"])
print(results.get("Outlier Test"))
print(results.get("Distortion Test"))
results["Main MR results"]
```

## Notes

- The module uses leave-one-out RSS for Global Test regardless of the Outlier Test flag.
- `NbDistribution` must exceed the number of complete SNP rows.
- Empirical P-values are simulation-based; zero counts are reported as an upper-bound string.
- Run `pytest -q` in the repository root to execute regression tests.
